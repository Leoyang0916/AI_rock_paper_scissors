```python
# rock_paper_scissors_basic
import random
from collections import Counter

#拳法對應表
ACTIONS = ["rock", "paper", "scissors"]
BEATS = {"rock": "scissors", "paper": "rock", "scissors": "paper"}

def result(my, opp):
    if my == opp: return 0 #平手
    return 1 if BEATS[my] == opp else -1

#選能刻對手的拳
def counter_move(opp_most_likely):
    for a, b in BEATS.items():
        if b == opp_most_likely:
            return a
    return random.choice(ACTIONS)

def main(rounds=20):
    print("Rock-Paper-Scissors (basic). 輸入 r/p/s，或按 Enter 由電腦代打。")
    hist = Counter()
    score = 0
    for t in range(1, rounds + 1):
        s = input(f"\nRound {t} 你的出拳(r/p/s/空白隨機): ").strip().lower()
        human = {"r": "rock", "p": "paper", "s": "scissors"}.get(s, random.choice(ACTIONS))
        opp = human  
        hist[opp] += 1

        # AI 根據統計最常出現的拳法，出剋它的拳法
        likely = hist.most_common(1)[0][0] if hist else random.choice(ACTIONS)
        ai = counter_move(likely)

        r = result(ai, opp)
        score += r
        outcome = "AI贏" if r == 1 else ("平手" if r == 0 else "AI輸")
        print(f"你: {opp:9s} ｜ AI: {ai:9s} ｜ 結果: {outcome} ｜ 累積分數(勝+1/負-1): {score}")

if __name__ == "__main__":
    main()
```

**核心思維**

這個程式主要是透過統計學來判斷最常出的拳，AI再判斷下一拳要出什麼，以下是基本流程:

玩家出拳 → 記錄在 Counter() → 分析統計 → 找出最常出的拳 →  AI 出能剋它的拳 →  比對結果、更新分數 →  顯示戰況、進入下一輪
